## Задание

Необходимо вычислить значение выражения:
$$A=\frac{sin(X)⋅lg(Y)}{1!}+\frac{ln(X)^3⋅cos(Y^2)}{3!}-\frac{sin(X^5)⋅lg(Y^3)}{5!}+\frac{ln(X^7)⋅cos(Y^4)}{7!}+...$$
Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $X$ и $Y$. Переменная, которая должна быть вычислена на основе независимых: $A$.

Следует обратить внимание, что в выражении неизвестное число слагаемых, поэтому необходимо ввести переменную $N$ для представления количества слагаемых.

В числителе чередуются синусы, косинусы и логарифмы в зависимости от порядкового номера шага. В знаменателе стоят факториалы нечетных чисел. Знаки чередуются (сложение и вычитание), а степень зависит от переменной, у $X$ это нечетные числа, а у $Y$ натуральные.

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $Z$:

```python

import math
A = 0
X = 4
Y = 3.55
n = 6
for i in range(1, n + 1):
    dx = 2 * i - 1
    z = math.factorial(dx)

    if i % 2 != 0:
        s = -1
        c = math.sin(X**dx) * math.log10(Y**i)
    else:
        s = 1
        c = math.log(X**dx) * math.cos(Y**i)

    A += s * c / z
print (A)
```
