## Задание

В файле pr5.txt содержится последовательность целых чисел.
Элементы последовательности могут принимать целые значения от
–10 000 до 10 000 включительно. Определите количество троек, в
которых хотя бы два из трёх элементов меньше, чем среднее
арифметическое всех чисел в файле, и десятичная запись хотя бы
двух из трёх элементов оканчивается на 8. В ответе запишите два
числа: сначала количество найденных троек, а затем –
максимальную сумму элементов таких троек. В данной задаче под
тройкой подразумевается три идущих подряд элемента
последовательности.

## Решение

Файл pr5.txt расположен в репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

````python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract5/pr5.txt
````

Реализуем чтение строк из файла и конвертацию отдельных строк в числовые значения. Каждое получаемое число будем добавлять в список `data`.

````python
data = []
with open('pr5.txt') as F:
  for line in F:
    digit = int(line)
    data.append(digit)

print(len(data))
print(data[:10])
````

Теперь ответим на поставленные вопросы использую список `data`

````python
sr = sum(data) / len(data)
print(sr)
````

````python
sr = sum(data) / len(data)
counter = 0
max_sum = None
for i in range(len(data) - 2):
    triple = data[i], data[i+1], data[i+2]
    less_cnt = sum(1 for v in triple if v < sr)
    end8_cnt = sum(1 for v in triple if str(v)[-1] == '8')

    if less_cnt >= 2 and end8_cnt >= 2:
        counter += 1
        s = sum(triple)
        if max_sum is None or s > max_sum:
            max_sum = s

print(f'Количество подходящих троек = {counter}')
print(f'Максимальная сумма = {max_sum}')
````

5. Опишите приведенные ниже списковые методы:

а) index() — возвращает индекс первого вхождения указанного значения в списке. Синтаксис: list.index(value, start, end). Если значения нет — ValueError.

б) insert() — вставляет элемент в список по указанной индексной позиции. Синтаксис: list.insert(index, value). Элементы, начиная с этой позиции, сдвигаются вправо.

в) sort() — сортирует элементы списка на месте (по возрастанию по умолчанию). Синтаксис: list.sort(key=None, reverse=False). Возвращает None; исходный список изменяется.

г) reverse() — изменяет порядок элементов списка на обратный (на месте). Синтаксис: list.reverse(). Возвращает None; исходный список изменяется.
