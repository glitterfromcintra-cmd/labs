## Задание

Напишите функцию, которая принимает в качестве параметра одно
натуральное число $N$ и возвращает сумму четных чисел на отрезке
$[1; N]$.

## Решение

### Объявление функции

```python
def s(n):
    a = 0
    for i in range(1, n + 1):
        if i % 2 == 0:
            a += i
    return a
```

### Демонстрация использования функции

```python
print(s(10))
print(s(3))
print(s(33))
```
